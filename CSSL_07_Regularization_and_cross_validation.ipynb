# Utility functions

```python
# Compute RMSE
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))
```

# Preprocessing
We need to find if there are missing values in the dataset and figure out a way to deal with them.

```python
rows_with_na = df[df.isna().any(axis=1)]
print("\nRows containing NAs:")
rows_with_na
len(rows_with_na)
```

Let's drop the rows with NaN values and split the dataset into train and test.

```python
df = df.dropna().reset_index(drop=True) #index is serial number of rows - unique number matched to each row - when removing rows, we have to reset the indices so that the order of 0 to n is balanced
target = "Salary" #want to drop this!
X = df.drop(columns=[target])
y = df[target]
print(X.shape, y.shape) #19 columns, 263 rows, y values correspond to the rows

# First split: train+val vs test (80% / 20%)
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, test_size=0.2, random_state=10
) #splitting it into train + validation dataa = 80% and the test = 20%

# Second split: train vs validation (75% / 25% of remaining 80%)
# This gives: 60% train, 20% val, 20% test overall
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size=0.25, random_state=20
) #why put 25%? so when i take my 80% of training, and then split that into validation - i want to put 20% as my validation data. 20% out of the 80% of my total training data is 1/4 or 25%.
#so when looking at the training + validation data: 25% is validation, 75% is training - but for the entire data = train + validation + test = it's actually 60% training, 20% test and 20% validation

print("Train size:", X_train.shape[0])
print("Validation size:", X_val.shape[0])
print("Test size:", X_test.shape[0])
```

Now we need to transform the categorical values into numerical values and also scale all the features so that they roughly lie within the same range. If some features are represented using large numbers while others are represented in small numbers, it can skew the error calculcations, similar to what happened when we did polynomial regression.

```python
numerical_cols = X_train.select_dtypes(include=[np.number]).columns.tolist() #selecting all features that have numbers = numerical features to be scaled
categorical_cols = X_train.select_dtypes(exclude=[np.number]).columns.tolist() #similarly for categorical features

print(numerical_cols)
print(categorical_cols)

#each feature is a column in the dataset.

preprocess = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols), # Scales columns containing numerical data using (x-mu)/sigma so that the values have mean 0 and standard deviation 1
        ("cat", OneHotEncoder(drop="first"), categorical_cols) # Encodes categorical values using one hot encoding
    ]
)

#checking the number
len(numerical_cols)
len(categorical_cols)
```
